```python
def simularSenalEEG(duracion=10, fs=256, num_canales=1, incluir_bandas=['alpha', 'beta', 'theta', 'delta', 'gamma'], ruido=0.1, semilla=None):
    """
    Simular una señal EEG realista con múltiples bandas de frecuencia.
    
    Las señales EEG se componen de diferentes bandas de frecuencia, cada una asociada
    con diferentes estados cerebrales:
    - Delta (0.5-4 Hz): Sueño profundo
    - Theta (4-8 Hz): Meditación, somnolencia
    - Alpha (8-13 Hz): Relajación, ojos cerrados
    - Beta (13-30 Hz): Concentración, actividad mental
    - Gamma (30-100 Hz): Procesamiento cognitivo de alto nivel
    
    Parámetros
    ----------
    duracion : float
        Duración de la señal en segundos (por defecto: 10)
    fs : int
        Frecuencia de muestreo en Hz (por defecto: 256)
    num_canales : int
        Número de canales EEG a simular (por defecto: 1)
    incluir_bandas : list
        Lista de bandas de frecuencia a incluir. Opciones: 'delta', 'theta', 'alpha', 'beta', 'gamma'
        (por defecto: todas las bandas)
    ruido : float
        Nivel de ruido a agregar a la señal (por defecto: 0.1)
    semilla : int o None
        Semilla para el generador de números aleatorios (por defecto: None)
    
    Devuelve
    -------
    t : numpy.ndarray
        Vector de tiempo en segundos
    senal : numpy.ndarray
        Señal EEG simulada. Si num_canales=1, devuelve un array 1D.
        Si num_canales>1, devuelve un array 2D de forma (num_canales, num_muestras)
    info_bandas : dict
        Diccionario con información sobre las bandas de frecuencia utilizadas
    
    Ejemplo
    -------
    >>> t, senal, info = simularSenalEEG(duracion=5, fs=256, num_canales=2)
    >>> print(f"Forma de la señal: {senal.shape}")
    >>> plt.plot(t, senal[0])  # Graficar el primer canal
    """
    
    import numpy as np
    
    # Configurar semilla para reproducibilidad
    if semilla is not None:
        np.random.seed(semilla)
    
    # Definir las bandas de frecuencia EEG (en Hz)
    bandas_freq = {
        'delta': (0.5, 4),    # Sueño profundo
        'theta': (4, 8),      # Somnolencia, meditación
        'alpha': (8, 13),     # Relajación
        'beta': (13, 30),     # Concentración
        'gamma': (30, 50)     # Procesamiento cognitivo
    }
    
    # Amplitudes típicas de cada banda (en microvolts)
    bandas_amp = {
        'delta': 75,
        'theta': 40,
        'alpha': 50,
        'beta': 20,
        'gamma': 10
    }
    
    # Crear el vector de tiempo
    num_muestras = int(duracion * fs)
    t = np.linspace(0, duracion, num_muestras)
    
    # Inicializar la señal
    if num_canales == 1:
        senal = np.zeros(num_muestras)
    else:
        senal = np.zeros((num_canales, num_muestras))
    
    # Información sobre las bandas utilizadas
    info_bandas = {}
    
    # Generar cada canal
    for canal in range(num_canales):
        senal_canal = np.zeros(num_muestras)
        
        # Agregar cada banda de frecuencia
        for banda in incluir_bandas:
            if banda in bandas_freq:
                freq_min, freq_max = bandas_freq[banda]
                amplitud_base = bandas_amp[banda]
                
                # Generar 2-3 componentes de frecuencia dentro de cada banda
                num_componentes = np.random.randint(2, 4)
                
                for _ in range(num_componentes):
                    # Frecuencia aleatoria dentro de la banda
                    freq = np.random.uniform(freq_min, freq_max)
                    
                    # Amplitud con variación aleatoria
                    amplitud = amplitud_base * np.random.uniform(0.5, 1.5)
                    
                    # Fase aleatoria
                    fase = np.random.uniform(0, 2 * np.pi)
                    
                    # Agregar componente sinusoidal
                    senal_canal += amplitud * np.sin(2 * np.pi * freq * t + fase)
                
                # Guardar información de la banda
                if banda not in info_bandas:
                    info_bandas[banda] = {
                        'rango_frecuencia': (freq_min, freq_max),
                        'amplitud_tipica': amplitud_base
                    }
        
        # Agregar ruido
        if ruido > 0:
            ruido_senal = np.random.normal(0, ruido * np.std(senal_canal), num_muestras)
            senal_canal += ruido_senal
        
        # Asignar el canal a la señal
        if num_canales == 1:
            senal = senal_canal
        else:
            senal[canal, :] = senal_canal
    
    return t, senal, info_bandas
```
